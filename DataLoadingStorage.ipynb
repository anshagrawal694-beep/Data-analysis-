CONVERSION OF TXT FILE IMTO CSV FILE

```python
# CREATE TEXT FILE , DATA ENTER .TXT FILE | WRITE OPERATIONS
with open("StudentsData.txt", "w") as f:
    f.write("student_name,roll_no,marks,attendance\n")
    f.write("A,101,85,Present\n")
    f.write("B,102,72,Absent\n")
    f.write("C,103,91,Present\n")
    f.write("D,104,68,Present\n")
```

```python
# CONVERSION OF TXT FILE IMTO CSV FILE
with open("StudentsData.txt", "r") as file:
  data = file.read()
print(data)
```

SQLITE

```python
import sqlite3
conn = sqlite3.connect("Students.db")
```

```python
cursor = conn.cursor()
```

```python
print("database created succsfully")
```

```python
cursor.execute("""CREATE TABLE IF NOT EXISTS STUDENT_TABLE( NAME TEXT,ROLL NO INTEGER , MARKS INTEGER)""")
print("table in created succsfully")
```

```python
#write the data into table
cursor.execute("INSERT INTO STUDENT_TABLE VALUES(?,?,?)",("Amit",101,90))
```

```python
conn.commit()
```

```python
conn.close()
```

```python
#Retrive data | read the data
conn = sqlite3.connect("Students.db")
cursor = conn.cursor()
cursor.execute("SELECT * FROM STUDENT_TABLE")
rows = cursor.fetchall()
for row in rows:
  print(row)
```

```python
#insert 4th record
cursor.execute("INSERT INTO STUDENT_TABLE VALUES(?,?,?)",("Ansh",1001,99))
conn.commit()
```

```python
#Retrive data again after inserting another row
#read the data again after inserting another row
conn = sqlite3.connect("Students.db")
cursor = conn.cursor()
cursor.execute("SELECT * FROM STUDENT_TABLE")
rows = cursor.fetchall()
for row in rows:
  print(row)
```

```python
conn.close()
```
