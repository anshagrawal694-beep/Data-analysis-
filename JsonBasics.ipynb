```python
# using this json text
import json
data = '''
[
    {"name": "Laptop", "price": 50000, "qty": 2},
    {"name": "Mouse", "price": 500, "qty": 3},
    {"name": "Keyboard", "price": 1000, "qty": 4},
    {"name": "Monitor", "price": 12000, "qty": 1}
]
'''
```

```python
# 1. Load JSON into a Python object and print its type
items = json.loads(data)
print(type(items))
```

```python
# 2. Print the name of the second item and the number of items
print("Second item:", items[1]["name"])
print("Number of items:", len(items))
```

```python
# 3. Print the value of owner. What Python value is it?
owner = None
print("Owner:", owner)
print("Python type:", type(owner))
```

```python
# 4. Convert the object back to JSON text with indentation 2
json_text = json.dumps(items, indent=2)
print(json_text)
```
