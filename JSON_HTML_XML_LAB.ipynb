```python
#1
import pandas as pd
data = [
    [101, "Ansh", 19, "CSE", "Mathura"],
    [102, "Riyan", 20, "ECE", "Agra"],
    [103, "anman", 19, "cse", "Delhi"],
    [104, "harsh", 21, "cse", "Jaipur"],
    [105, "Rahul", 20, "CSE", "Noida"]
]
pd.DataFrame(data).to_csv("students.csv", index=False, header=False)
df = pd.read_csv(
    "students.csv",
    header=None,
    names=["ID", "Name", "Age", "Branch", "City"],
    index_col=0
)
print(df)
```

```python
#2
import pandas as pd
data = {"Name": ["Ansh", "kapil", "Aman"],"Marks": [90, None, 85]}
df = pd.DataFrame(data)
df.to_csv("marks.txt", sep="|", index=False, na_rep="-999")
df2 = pd.read_csv("marks.txt", sep="|", na_values=["-999"])
print(df2)
```

```python
#3
import pandas as pd
means = []
for chunk in pd.read_csv("/content/Titanic-Dataset.csv",chunksize=2000):
    means.append(chunk["Age"].mean())
print("Mean of each chunk:", means)
```

```python
#4
import pandas as pd
df = pd.DataFrame({
    "Name": ["Ansh", "harsh", "kapil"],
    "Age": [19, 20, 19],
    "Branch": ["CSE", "cse", "cse"]
})
df.to_json("students.json", orient="records")
df2 = pd.read_json("students.json", orient="records")
print(df.equals(df2))
```
