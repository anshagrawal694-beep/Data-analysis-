***SORTING AND GROUPING***

```python
df.sort_values(by='Age')
```

```python
# Sort by a single column (e.g., 'Age') in ascending order (default)
print("\nDataFrame sorted by 'Age' (ascending):")
display(df.sort_values(by='Age').head())

# Sort by a single column (e.g., 'Glucose') in descending order
print("\nDataFrame sorted by 'Glucose' (descending):")
display(df.sort_values(by='Glucose', ascending=False).head())

# Sort by multiple columns (e.g., 'Age' then 'Glucose')
# 'Age' in ascending, 'Glucose' in descending
print("\nDataFrame sorted by 'Age' (asc) and 'Glucose' (desc):")
display(df.sort_values(by=['Age', 'Glucose'], ascending=[True, False]).head())
```

```python
# Get value counts for a column (e.g., 'Age') and sort them
print("\nValue counts for 'Age' (sorted by count, descending):\n")
display(df['Age'].value_counts().sort_values(ascending=False).head())

# You can also sort by the index (the age values themselves)
print("\nValue counts for 'Age' (sorted by age value, ascending):\n")
display(df['Age'].value_counts().sort_index().head())
```
